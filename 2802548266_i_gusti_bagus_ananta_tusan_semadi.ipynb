```python
def my_thermo_stat(temp, desired_temp):
  if temp < desired_temp - 5:
      status = 'Heat'
  elif temp > desired_temp + 5:
      status = 'AC'
  else:
      status = 'off'
  return status
```

```python
status = my_thermo_stat(65,75)
print(status)
```

```python
status = my_thermo_stat(75,65)
print(status)
```

```python
status = my_thermo_stat(65,63)
print(status)
```

```python
def login():
    username = input("Username: ")
    password = input("Password: ")

    # Validasi username dan password
    if password != "Bn123":
        print("Username atau Password salah. Coba lagi.")
        return False
    return True

def calculate_final_score(uas, uts, asg):
    # Menghitung nilai akhir
    final_score = (0.5 * uas) + (0.3 * uts) + (0.2 * asg)
    return final_score

def determine_grade(final_score):
    # Menentukan grade berdasarkan nilai akhir
    if final_score >= 85:
        return 'A'
    elif final_score >= 70:
        return 'B'
    elif final_score >= 55:
        return 'C'
    elif final_score >= 40:
        return 'D'
    else:
        return 'E'

# Program utama
def main():
    # Login
    while not login():
        pass

    print("Login berhasil!")

    # Input nilai
    nama = input("Nama Siswa: ")

    while True:
        try:
            uas = int(input("Nilai UAS: "))
            uts = int(input("Nilai UTS: "))
            asg = int(input("Nilai ASG: "))
            break
        except ValueError:
            print("Harap masukkan angka yang valid!")

    # Menghitung nilai akhir
    final_score = calculate_final_score(uas, uts, asg)

    # Menentukan grade
    grade = determine_grade(final_score)

    # Menampilkan hasil
    print(f"\nNama Siswa: {nama}")
    print(f"Nilai UAS: {uas}")
    print(f"Nilai UTS: {uts}")
    print(f"Nilai ASG: {asg}")
    print(f"Nilai Akhir: {final_score:.2f}")
    print(f"Grade: {grade}")

if __name__ == "__main__":
    main()
```
